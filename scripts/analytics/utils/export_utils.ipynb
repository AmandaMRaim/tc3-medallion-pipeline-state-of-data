```
from pathlib import Path


def exportar_csv(df, caminho_saida, nome_arquivo):
    """
    Exporta um DataFrame Spark para CSV.
    """

    caminho_saida = Path(caminho_saida)

    caminho_saida.mkdir(
        parents=True,
        exist_ok=True
    )

    caminho_arquivo = (
        caminho_saida
        / nome_arquivo
    )

    df.toPandas().to_csv(
        caminho_arquivo,
        index=False
    )

    print(f"Arquivo exportado: {caminho_arquivo}")
```
