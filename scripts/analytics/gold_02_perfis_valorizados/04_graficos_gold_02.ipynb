```
# ---------------------------------------------------------------------
# CONFIGURAÇÕES INICIAIS
# ---------------------------------------------------------------------

from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt


# Caminhos
SCRIPT_DIR = Path(__file__).resolve().parent
PROJECT_ROOT = Path(__file__).resolve().parents[3]

OUTPUT_DIR = PROJECT_ROOT / "scripts" / "Analytics" / "outputs" / "gold_02"
GRAFICOS_DIR = PROJECT_ROOT / "scripts" / "Analytics" / "graficos" / "gold_02"

GRAFICOS_DIR.mkdir(parents=True, exist_ok=True)


# ---------------------------------------------------------------------
# CARREGAR OS CSVs
# ---------------------------------------------------------------------

caminho_atual = OUTPUT_DIR / "perfis_valorizados_atual.csv"
caminho_historico = OUTPUT_DIR / "perfis_evolucao_historica.csv"

if not caminho_atual.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {caminho_atual}")

if not caminho_historico.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {caminho_historico}")

df_atual = pd.read_csv(caminho_atual)
df_historico = pd.read_csv(caminho_historico)

print("\nARQUIVOS CARREGADOS:")
print(f"- {caminho_atual}")
print(f"- {caminho_historico}")


# ---------------------------------------------------------------------
# CONFIGURAÇÕES DE APOIO
# ---------------------------------------------------------------------

nomes_cargos = {
    "Cientista de Dados/Data Scientist": "Cientista de Dados",
    "Engenheiro de Machine Learning/ML Engineer/AI Engineer": "ML / AI Engineer",
    "Engenharia e Arquitetura de Dados": "Engenharia e Arquitetura",
    "Analytics Engineer": "Analytics Engineer",
    "Desenvolvedor/ Engenheiro de Software/ Analista de Sistemas": "Desenvolvimento / Eng. Software",
    "Analista de Dados/Data Analyst": "Analista de Dados",
    "Analista de BI/BI Analyst": "Analista de BI",
    "Analista de Negócios/Business Analyst": "Analista de Negócios",
    "Analista de Suporte/Analista Técnico": "Analista de Suporte"
}

rotulos_faixas = {
    1: "< R$ 1 mil",
    2: "R$ 1–2 mil",
    3: "R$ 2–3 mil",
    4: "R$ 3–4 mil",
    5: "R$ 4–6 mil",
    6: "R$ 6–8 mil",
    7: "R$ 8–12 mil",
    8: "R$ 12–16 mil",
    9: "R$ 16–20 mil",
    10: "R$ 20–25 mil",
    11: "R$ 25–30 mil",
    12: "R$ 30–40 mil",
    13: "> R$ 40 mil"
}

rotulos_faixas_executivo = {
    "Menos de R$ 1.000/mês": "< R$ 1 mil",
    "de R$ 1.001/mês a R$ 2.000/mês": "R$ 1–2 mil",
    "de R$ 2.001/mês a R$ 3.000/mês": "R$ 2–3 mil",
    "de R$ 3.001/mês a R$ 4.000/mês": "R$ 3–4 mil",
    "de R$ 4.001/mês a R$ 6.000/mês": "R$ 4–6 mil",
    "de R$ 6.001/mês a R$ 8.000/mês": "R$ 6–8 mil",
    "de R$ 8.001/mês a R$ 12.000/mês": "R$ 8–12 mil",
    "de R$ 12.001/mês a R$ 16.000/mês": "R$ 12–16 mil",
    "de R$ 16.001/mês a R$ 20.000/mês": "R$ 16–20 mil",
    "de R$ 20.001/mês a R$ 25.000/mês": "R$ 20–25 mil",
    "de R$ 25.001/mês a R$ 30.000/mês": "R$ 25–30 mil",
    "de R$ 30.001/mês a R$ 40.000/mês": "R$ 30–40 mil",
    "Acima de R$ 40.001/mês": "> R$ 40 mil"
}

ordem_niveis = {
    "Júnior": 1,
    "Pleno": 2,
    "Sênior": 3,
    "Especialista/Staff+": 4
}

edicoes = ["2023-2024", "2024-2025", "2025-2026"]

ordem_edicoes = {
    "2023-2024": 0,
    "2024-2025": 1,
    "2025-2026": 2
}


# ---------------------------------------------------------------------
# PREPARAR BASE ATUAL
# ---------------------------------------------------------------------

df_atual["cargo_exibicao"] = df_atual["cargo_harmonizado"].replace(nomes_cargos)
df_atual["ordem_nivel"] = df_atual["nivel"].map(ordem_niveis)

df_atual["ranking"] = pd.to_numeric(df_atual["ranking"])
df_atual["ordem_faixa_mediana"] = pd.to_numeric(df_atual["ordem_faixa_mediana"])
df_atual["ordem_faixa_p75"] = pd.to_numeric(df_atual["ordem_faixa_p75"])
df_atual["total_respondentes"] = pd.to_numeric(df_atual["total_respondentes"])

# Perfis em primeiro lugar
df_lideres = df_atual[df_atual["ranking"] == 1].copy()

df_lideres["p50_exibicao"] = (
    df_lideres["faixa_salarial_mediana"].replace(rotulos_faixas_executivo)
)

df_lideres["p75_exibicao"] = (
    df_lideres["faixa_salarial_p75"].replace(rotulos_faixas_executivo)
)

df_lideres = (
    df_lideres
    .sort_values(["ordem_nivel", "cargo_exibicao"])
    .reset_index(drop=True)
)


# ---------------------------------------------------------------------
# PERFIS COM MAIORES FAIXAS SALARIAIS - 2025-2026
# ---------------------------------------------------------------------

fig, ax = plt.subplots(figsize=(16, 8))
ax.axis("off")

fig.suptitle(
    "Perfis com maiores faixas salariais por senioridade — 2025-2026",
    x=0.06,
    y=0.97,
    ha="left",
    fontsize=20,
    fontweight="bold"
)

fig.text(
    0.06,
    0.91,
    (
        "ML/AI Engineer lidera nos níveis Sênior e Especialista/Staff+; "
        "no Pleno, três perfis compartilham a liderança salarial."
    ),
    ha="left",
    fontsize=13
)

x_nivel = 0.06
x_perfil = 0.25
x_amostra = 0.61
x_p50 = 0.72
x_p75 = 0.86

ax.text(
    x_nivel, 0.86, "Senioridade",
    transform=ax.transAxes, fontsize=11, fontweight="bold"
)

ax.text(
    x_perfil, 0.86, "Perfil",
    transform=ax.transAxes, fontsize=11, fontweight="bold"
)

ax.text(
    x_amostra, 0.86, "Amostra",
    transform=ax.transAxes, fontsize=11, fontweight="bold"
)

ax.text(
    x_p50, 0.86, "P50",
    transform=ax.transAxes, fontsize=11, fontweight="bold"
)

ax.text(
    x_p75, 0.86, "P75",
    transform=ax.transAxes, fontsize=11, fontweight="bold"
)

ax.plot([0.05, 0.96], [0.835, 0.835], transform=ax.transAxes, linewidth=1)

inicio_y = 0.77
espacamento = 0.105
nivel_anterior = None

for i, linha in df_lideres.iterrows():
    y = inicio_y - (i * espacamento)

    if nivel_anterior is not None and linha["nivel"] != nivel_anterior:
        ax.plot(
            [0.05, 0.96],
            [y + 0.055, y + 0.055],
            transform=ax.transAxes,
            linewidth=0.7,
            alpha=0.35
        )

    if linha["nivel"] != nivel_anterior:
        ax.text(
            x_nivel,
            y,
            linha["nivel"],
            transform=ax.transAxes,
            fontsize=11,
            fontweight="bold",
            va="center"
        )

    ax.text(
        x_perfil,
        y,
        linha["cargo_exibicao"],
        transform=ax.transAxes,
        fontsize=11,
        va="center"
    )

    ax.text(
        x_amostra,
        y,
        f'n = {int(linha["total_respondentes"])}',
        transform=ax.transAxes,
        fontsize=11,
        va="center"
    )

    ax.text(
        x_p50,
        y,
        linha["p50_exibicao"],
        transform=ax.transAxes,
        fontsize=11,
        fontweight="bold",
        va="center"
    )

    ax.text(
        x_p75,
        y,
        linha["p75_exibicao"],
        transform=ax.transAxes,
        fontsize=11,
        fontweight="bold",
        va="center"
    )

    nivel_anterior = linha["nivel"]

fig.text(
    0.06,
    0.105,
    (
        "P50 = faixa salarial mediana | "
        "P75 = faixa em que 75% da distribuição salarial está acumulada"
    ),
    ha="left",
    fontsize=10,
    fontweight="bold"
)

fig.text(
    0.06,
    0.055,
    (
        "Nota: remuneração utilizada como proxy de valorização. "
        "Comparações realizadas dentro da mesma senioridade e somente para perfis com n ≥ 20. "
        "Empates são preservados quando P50 e P75 são iguais."
    ),
    ha="left",
    fontsize=9
)

caminho_grafico_01 = GRAFICOS_DIR / "01_perfis_valorizados_atual.png"

plt.savefig(caminho_grafico_01, dpi=300, bbox_inches="tight")
plt.show()

print(f"\nGráfico salvo em: {caminho_grafico_01}")


# ---------------------------------------------------------------------
# PREPARAR BASE HISTÓRICA
# ---------------------------------------------------------------------

df_historico["cargo_exibicao"] = (
    df_historico["cargo_harmonizado"].replace(nomes_cargos)
)

df_historico["ordem_edicao"] = df_historico["edicao"].map(ordem_edicoes)
df_historico["ordem_faixa_p50"] = pd.to_numeric(df_historico["ordem_faixa_p50"])
df_historico["ordem_faixa_p75"] = pd.to_numeric(df_historico["ordem_faixa_p75"])

df_historico["perfil"] = (
    df_historico["cargo_exibicao"]
    + " | "
    + df_historico["nivel"]
)

df_historico = (
    df_historico
    .sort_values(["perfil", "ordem_edicao"])
    .reset_index(drop=True)
)


# ---------------------------------------------------------------------
# EVOLUÇÃO DA FAIXA SALARIAL MEDIANA - P50
# ---------------------------------------------------------------------

fig, ax = plt.subplots(figsize=(17, 9))

for perfil, dados in df_historico.groupby("perfil"):
    dados = dados.sort_values("ordem_edicao")

    linha, = ax.plot(
        dados["ordem_edicao"],
        dados["ordem_faixa_p50"],
        marker="o",
        linewidth=2.8,
        markersize=7
    )

    ultimo = dados.iloc[-1]

    ax.annotate(
        perfil,
        xy=(ultimo["ordem_edicao"], ultimo["ordem_faixa_p50"]),
        xytext=(12, 0),
        textcoords="offset points",
        va="center",
        fontsize=10.5,
        color=linha.get_color()
    )

menor_p50 = int(df_historico["ordem_faixa_p50"].min())
maior_p50 = int(df_historico["ordem_faixa_p50"].max())

faixas_p50 = list(range(menor_p50, maior_p50 + 1))

ax.set_yticks(faixas_p50)

ax.set_yticklabels(
    [rotulos_faixas[x] for x in faixas_p50],
    fontsize=12
)

ax.set_xticks([0, 1, 2])
ax.set_xticklabels(edicoes, fontsize=12)

ax.set_xlim(-0.10, 2.90)

fig.suptitle(
    "Evolução da faixa salarial mediana (P50) — 2023-2024 a 2025-2026",
    x=0.07,
    y=0.97,
    ha="left",
    fontsize=20,
    fontweight="bold"
)

ax.set_title(
    (
        "A maior parte dos perfis permaneceu estável; "
        "Analytics Engineer Pleno e Desenvolvimento Júnior avançaram uma faixa."
    ),
    loc="left",
    fontsize=13,
    pad=20
)

ax.set_xlabel("Edição da pesquisa", fontsize=12)
ax.set_ylabel("Faixa salarial mediana (P50)", fontsize=12)

ax.grid(axis="y", alpha=0.20)
ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

fig.text(
    0.07,
    0.025,
    (
        "Nota: comparação nominal entre faixas salariais. "
        "Os valores não foram corrigidos por inflação. "
        "Somente combinações cargo + senioridade com n ≥ 20 nas três edições foram consideradas."
    ),
    ha="left",
    fontsize=10
)

plt.tight_layout(rect=[0.05, 0.08, 0.96, 0.91])

caminho_grafico_02 = GRAFICOS_DIR / "02_evolucao_p50.png"

plt.savefig(caminho_grafico_02, dpi=300, bbox_inches="tight")
plt.show()

print(f"\nGráfico salvo em: {caminho_grafico_02}")


# ---------------------------------------------------------------------
# EVOLUÇÃO DO TOPO DA DISTRIBUIÇÃO - P75
# ---------------------------------------------------------------------

fig, ax = plt.subplots(figsize=(17, 9))

# Ajustes para evitar sobreposição de rótulos
ajuste_rotulos_p75 = {
    "Analytics Engineer | Pleno": 12,
    "Cientista de Dados | Júnior": -12
}

for perfil, dados in df_historico.groupby("perfil"):
    dados = dados.sort_values("ordem_edicao")

    linha, = ax.plot(
        dados["ordem_edicao"],
        dados["ordem_faixa_p75"],
        marker="o",
        linewidth=2.8,
        markersize=7
    )

    ultimo = dados.iloc[-1]
    deslocamento_y = ajuste_rotulos_p75.get(perfil, 0)

    ax.annotate(
        perfil,
        xy=(ultimo["ordem_edicao"], ultimo["ordem_faixa_p75"]),
        xytext=(12, deslocamento_y),
        textcoords="offset points",
        va="center",
        fontsize=10.5,
        color=linha.get_color()
    )

menor_p75 = int(df_historico["ordem_faixa_p75"].min())
maior_p75 = int(df_historico["ordem_faixa_p75"].max())

faixas_p75 = list(range(menor_p75, maior_p75 + 1))

ax.set_yticks(faixas_p75)

ax.set_yticklabels(
    [rotulos_faixas[x] for x in faixas_p75],
    fontsize=12
)

ax.set_xticks([0, 1, 2])
ax.set_xticklabels(edicoes, fontsize=12)

ax.set_xlim(-0.10, 2.90)

fig.suptitle(
    "Evolução do topo da distribuição salarial (P75) — 2023-2024 a 2025-2026",
    x=0.07,
    y=0.97,
    ha="left",
    fontsize=20,
    fontweight="bold"
)

ax.set_title(
    (
        "ML/AI Engineer Sênior apresentou o maior avanço: "
        "o P75 passou de R$ 20–25 mil para R$ 30–40 mil."
    ),
    loc="left",
    fontsize=13,
    pad=20
)

ax.set_xlabel("Edição da pesquisa", fontsize=12)
ax.set_ylabel("Faixa salarial do P75", fontsize=12)

ax.grid(axis="y", alpha=0.20)
ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

fig.text(
    0.07,
    0.025,
    (
        "Nota: comparação nominal entre faixas salariais. "
        "Os valores não foram corrigidos por inflação. "
        "P75 representa a faixa em que 75% da distribuição salarial de cada perfil está acumulada."
    ),
    ha="left",
    fontsize=10
)

plt.tight_layout(rect=[0.05, 0.08, 0.96, 0.91])

caminho_grafico_03 = GRAFICOS_DIR / "03_evolucao_p75.png"

plt.savefig(caminho_grafico_03, dpi=300, bbox_inches="tight")
plt.show()

print(f"\nGráfico salvo em: {caminho_grafico_03}")
```
